# Grain types and other labeled data

For efficiency reasons, data variables across a large dataset are best handled as numeric codes. To ease interpreting numeric codes, `xsnow` implements strategies for translating these codes into human-readable labels. 

Grain types are a stereotypical example for this concept. While the International Classification for Seasonal Snow on the Ground lists grain type names (and acronyms) like *Precipitation Particles* (PP), *Faceted Crystals* (FC), and more, SNOWPACK uses a one-to-three-digit code to describe the *(primary grain shape, secondary grain shape, grain shape cycle)*.
